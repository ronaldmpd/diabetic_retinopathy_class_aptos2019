```python
# ================================================================
# EFFICIENTNET-B0 + PYTORCH + MPS
# APTOS 2019 BLINDNESS DETECTION
# APPLE SILICON M4
# ================================================================
#
# CLASES:
# 0 = No DR
# 1 = Mild
# 2 = Moderate
# 3 = Severe
# 4 = Proliferative DR
#
# ================================================================


# ================================================================
# 1. INSTALACIÓN DE LIBRERÍAS
# ================================================================
#
# Ejecutar una sola vez en la terminal:
#
# python -m pip install torch torchvision pandas numpy
# python -m pip install matplotlib seaborn scikit-learn pillow tqdm
#
# ================================================================


# ================================================================
# 2. IMPORTACIONES
# ================================================================

import os
import random
import copy

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image

from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import (
    Dataset,
    DataLoader
)

from torchvision import transforms

from torchvision.models import (
    efficientnet_b0,
    EfficientNet_B0_Weights
)

from sklearn.model_selection import train_test_split

from sklearn.utils.class_weight import compute_class_weight

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)


# ================================================================
# 3. CONFIGURACIÓN GENERAL
# ================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)


# ================================================================
# 4. CONFIGURACIÓN DEL DEVICE
# ================================================================

print("=" * 70)
print("CONFIGURACIÓN DEL SISTEMA")
print("=" * 70)

print(
    "Versión de PyTorch:",
    torch.__version__
)


if torch.backends.mps.is_available():

    device = torch.device("mps")

    print(
        "✓ Apple Silicon GPU MPS disponible."
    )

elif torch.cuda.is_available():

    device = torch.device("cuda")

    print(
        "✓ NVIDIA CUDA disponible."
    )

else:

    device = torch.device("cpu")

    print(
        "⚠ MPS/CUDA no disponible."
    )

    print(
        "Se utilizará CPU."
    )


print(
    "\nDevice seleccionado:",
    device
)


# ================================================================
# 5. RUTA DEL DATASET
# ================================================================

# ================================================================
# IMPORTANTE:
# MODIFICA ESTA RUTA
# ================================================================

BASE_DIR = (
   "./aptos2019-blindness-detection"
)


CSV_PATH = os.path.join(

    BASE_DIR,

    "train.csv"

)


IMAGE_DIR = os.path.join(

    BASE_DIR,

    "train_images"

)


# ================================================================
# 6. PARÁMETROS
# ================================================================

IMG_SIZE = 224

# Para Mac M4:
#
# 16 GB RAM  -> 16
# 24 GB RAM  -> 16/32
# 32 GB RAM+ -> 32
#
BATCH_SIZE = 16

NUM_CLASSES = 5

NUM_WORKERS = 0

INITIAL_EPOCHS = 10

FINE_TUNE_EPOCHS = 10

LEARNING_RATE = 1e-4

FINE_TUNE_LR = 1e-5

WEIGHT_DECAY = 1e-4

MODEL_PATH = (
    "best_efficientnet_b0_aptos_mps.pth"
)


# ================================================================
# 7. COMPROBAR DATASET
# ================================================================

print("\n")
print("=" * 70)
print("COMPROBANDO DATASET")
print("=" * 70)

print(
    "CSV:",
    CSV_PATH
)

print(
    "Directorio imágenes:",
    IMAGE_DIR
)


if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(

        f"\nNo se encontró el archivo:\n"
        f"{CSV_PATH}\n\n"
        f"Modifica BASE_DIR."

    )


if not os.path.exists(IMAGE_DIR):

    raise FileNotFoundError(

        f"\nNo se encontró el directorio:\n"
        f"{IMAGE_DIR}\n\n"
        f"Modifica BASE_DIR."

    )


print(
    "\n✓ Dataset encontrado."
)


# ================================================================
# 8. CARGAR CSV
# ================================================================

df = pd.read_csv(
    CSV_PATH
)


print("\n")
print("=" * 70)
print("INFORMACIÓN DEL DATASET")
print("=" * 70)

print(
    "Número de registros:",
    len(df)
)

print(
    "\nColumnas:"
)

print(
    df.columns.tolist()
)

print(
    "\nPrimeras filas:"
)

print(
    df.head()
)


# ================================================================
# 9. CREAR RUTA DE IMAGEN
# ================================================================

df["image_path"] = df[
    "id_code"
].apply(

    lambda x:
    os.path.join(

        IMAGE_DIR,

        str(x) + ".png"

    )

)


# ================================================================
# 10. COMPROBAR IMÁGENES
# ================================================================

df["exists"] = df[
    "image_path"
].apply(
    os.path.exists
)


print("\n")
print(
    "Imágenes encontradas:",
    df["exists"].sum()
)

print(
    "Imágenes faltantes:",
    (~df["exists"]).sum()
)


df = df[
    df["exists"]
].copy()


df.reset_index(
    drop=True,
    inplace=True
)


# ================================================================
# 11. NOMBRES DE LAS CLASES
# ================================================================

class_names = [

    "No DR",

    "Mild",

    "Moderate",

    "Severe",

    "Proliferative DR"

]


# ================================================================
# 12. DISTRIBUCIÓN DE CLASES
# ================================================================

print("\n")
print("=" * 70)
print("DISTRIBUCIÓN DE CLASES")
print("=" * 70)


distribution = (

    df["diagnosis"]
    .value_counts()
    .sort_index()

)


for class_id, count in distribution.items():

    print(

        f"Clase {class_id} - "
        f"{class_names[class_id]}: "
        f"{count}"

    )


# ================================================================
# 13. GRÁFICA DE DISTRIBUCIÓN
# ================================================================

plt.figure(
    figsize=(10, 6)
)


sns.countplot(

    data=df,

    x="diagnosis",

    hue="diagnosis",

    palette="viridis",

    legend=False

)


plt.title(
    "Distribución de clases - APTOS 2019"
)

plt.xlabel(
    "Grado de retinopatía"
)

plt.ylabel(
    "Número de imágenes"
)

plt.xticks(

    ticks=[0, 1, 2, 3, 4],

    labels=[

        "0 - No DR",

        "1 - Mild",

        "2 - Moderate",

        "3 - Severe",

        "4 - Proliferative"

    ]

)


plt.tight_layout()

plt.show()


# ================================================================
# 14. TRAIN / VALIDATION / TEST
# ================================================================

train_df, temp_df = train_test_split(

    df,

    test_size=0.30,

    stratify=df["diagnosis"],

    random_state=SEED

)


val_df, test_df = train_test_split(

    temp_df,

    test_size=0.50,

    stratify=temp_df["diagnosis"],

    random_state=SEED

)


print("\n")
print("=" * 70)
print("DIVISIÓN DEL DATASET")
print("=" * 70)

print(
    "Train:",
    len(train_df)
)

print(
    "Validation:",
    len(val_df)
)

print(
    "Test:",
    len(test_df)
)


# ================================================================
# 15. DATASET PERSONALIZADO
# ================================================================

class APTOSDataset(
    Dataset
):


    def __init__(
        self,
        dataframe,
        transform=None
    ):

        self.dataframe = (

            dataframe.reset_index(
                drop=True
            )

        )

        self.transform = transform


    def __len__(
        self
    ):

        return len(
            self.dataframe
        )


    def __getitem__(
        self,
        index
    ):

        row = self.dataframe.iloc[
            index
        ]


        image_path = row[
            "image_path"
        ]


        label = int(
            row[
                "diagnosis"
            ]
        )


        image = Image.open(

            image_path

        ).convert(
            "RGB"
        )


        if self.transform:

            image = self.transform(
                image
            )


        return (

            image,

            label

        )


# ================================================================
# 16. TRANSFORMACIONES
# ================================================================

# ================================================================
# TRAIN
# ================================================================

train_transform = transforms.Compose([

    transforms.Resize(

        (IMG_SIZE, IMG_SIZE)

    ),

    transforms.RandomHorizontalFlip(

        p=0.5

    ),

    transforms.RandomRotation(

        degrees=10

    ),

    transforms.RandomResizedCrop(

        IMG_SIZE,

        scale=(0.90, 1.0)

    ),

    transforms.ColorJitter(

        brightness=0.10,

        contrast=0.10

    ),

    transforms.ToTensor(),

    transforms.Normalize(

        mean=[

            0.485,

            0.456,

            0.406

        ],

        std=[

            0.229,

            0.224,

            0.225

        ]

    )

])


# ================================================================
# VALIDATION / TEST
# ================================================================

test_transform = transforms.Compose([

    transforms.Resize(

        (IMG_SIZE, IMG_SIZE)

    ),

    transforms.ToTensor(),

    transforms.Normalize(

        mean=[

            0.485,

            0.456,

            0.406

        ],

        std=[

            0.229,

            0.224,

            0.225

        ]

    )

])


# ================================================================
# 17. CREAR DATASETS
# ================================================================

train_dataset = APTOSDataset(

    train_df,

    transform=train_transform

)


val_dataset = APTOSDataset(

    val_df,

    transform=test_transform

)


test_dataset = APTOSDataset(

    test_df,

    transform=test_transform

)


# ================================================================
# 18. CREAR DATALOADERS
# ================================================================

train_loader = DataLoader(

    train_dataset,

    batch_size=BATCH_SIZE,

    shuffle=True,

    num_workers=NUM_WORKERS,

    pin_memory=False

)


val_loader = DataLoader(

    val_dataset,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=False

)


test_loader = DataLoader(

    test_dataset,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=False

)


print(
    "\n✓ DataLoaders creados."
)


# ================================================================
# 19. CLASS WEIGHTS
# ================================================================

classes = np.array(

    sorted(

        train_df[
            "diagnosis"
        ].unique()

    )

)


class_weights_np = compute_class_weight(

    class_weight="balanced",

    classes=classes,

    y=train_df[
        "diagnosis"
    ]

)


class_weights = torch.tensor(

    class_weights_np,

    dtype=torch.float32

).to(device)


print("\n")
print("=" * 70)
print("CLASS WEIGHTS")
print("=" * 70)


for i, weight in enumerate(
    class_weights_np
):

    print(

        f"{i} - "
        f"{class_names[i]}: "
        f"{weight:.4f}"

    )


# ================================================================
# 20. CARGAR EFFICIENTNET-B0
# ================================================================

print("\n")
print("=" * 70)
print("CARGANDO EFFICIENTNET-B0")
print("=" * 70)


weights = (
    EfficientNet_B0_Weights.DEFAULT
)


model = efficientnet_b0(

    weights=weights

)


print(
    "✓ EfficientNet-B0 cargado."
)


# ================================================================
# 21. MODIFICAR CLASIFICADOR
# ================================================================

# EfficientNet-B0 tiene:
#
# model.classifier[1]
#
# como capa Linear final.


num_features = (

    model.classifier[
        1
    ].in_features

)


model.classifier = nn.Sequential(

    nn.Dropout(

        p=0.40

    ),

    nn.Linear(

        num_features,

        256

    ),

    nn.ReLU(),

    nn.Dropout(

        p=0.30

    ),

    nn.Linear(

        256,

        NUM_CLASSES

    )

)


# ================================================================
# 22. ENVIAR MODELO A MPS
# ================================================================

model = model.to(
    device
)


print(
    "\nModelo enviado a:",
    device
)


# ================================================================
# 23. CONGELAR BACKBONE
# ================================================================

for param in model.parameters():

    param.requires_grad = False


# Entrenar solamente classifier
for param in model.classifier.parameters():

    param.requires_grad = True


# ================================================================
# 24. CONTAR PARÁMETROS
# ================================================================

def count_parameters(
    model
):

    total = sum(

        p.numel()

        for p in model.parameters()

    )


    trainable = sum(

        p.numel()

        for p in model.parameters()

        if p.requires_grad

    )


    return (

        total,

        trainable

    )


total_params, trainable_params = (

    count_parameters(
        model
    )

)


print("\n")
print(
    f"Parámetros totales: "
    f"{total_params:,}"
)

print(
    f"Parámetros entrenables: "
    f"{trainable_params:,}"
)


# ================================================================
# 25. FUNCIÓN DE PÉRDIDA
# ================================================================

criterion = nn.CrossEntropyLoss(

    weight=class_weights

)


# ================================================================
# 26. OPTIMIZADOR
# ================================================================

optimizer = optim.Adam(

    filter(

        lambda p:
        p.requires_grad,

        model.parameters()

    ),

    lr=LEARNING_RATE,

    weight_decay=WEIGHT_DECAY

)


# ================================================================
# 27. FUNCIÓN TRAIN
# ================================================================

def train_one_epoch(

    model,

    loader,

    criterion,

    optimizer,

    device

):


    model.train()


    running_loss = 0.0

    correct = 0

    total = 0


    progress_bar = tqdm(

        loader,

        desc="Training",

        leave=False

    )


    for images, labels in progress_bar:


        images = images.to(
            device
        )

        labels = labels.to(
            device
        )


        optimizer.zero_grad()


        outputs = model(
            images
        )


        loss = criterion(

            outputs,

            labels

        )


        loss.backward()


        optimizer.step()


        running_loss += (

            loss.item()

            *

            images.size(0)

        )


        _, predicted = torch.max(

            outputs,

            1

        )


        total += (

            labels.size(0)

        )


        correct += (

            predicted == labels

        ).sum().item()


        progress_bar.set_postfix(

            loss=f"{loss.item():.4f}"

        )


    epoch_loss = (

        running_loss /

        total

    )


    epoch_accuracy = (

        correct /

        total

    )


    return (

        epoch_loss,

        epoch_accuracy

    )


# ================================================================
# 28. FUNCIÓN VALIDATION
# ================================================================

def validate(

    model,

    loader,

    criterion,

    device

):


    model.eval()


    running_loss = 0.0

    correct = 0

    total = 0


    all_labels = []

    all_predictions = []


    with torch.no_grad():


        for images, labels in loader:


            images = images.to(
                device
            )

            labels = labels.to(
                device
            )


            outputs = model(
                images
            )


            loss = criterion(

                outputs,

                labels

            )


            running_loss += (

                loss.item()

                *

                images.size(0)

            )


            _, predicted = torch.max(

                outputs,

                1

            )


            total += (

                labels.size(0)

            )


            correct += (

                predicted == labels

            ).sum().item()


            all_labels.extend(

                labels.cpu().numpy()

            )


            all_predictions.extend(

                predicted.cpu().numpy()

            )


    epoch_loss = (

        running_loss /

        total

    )


    epoch_accuracy = (

        correct /

        total

    )


    return (

        epoch_loss,

        epoch_accuracy,

        np.array(
            all_labels
        ),

        np.array(
            all_predictions
        )

    )


# ================================================================
# 29. HISTORIAL
# ================================================================

history = {

    "train_loss": [],

    "train_acc": [],

    "val_loss": [],

    "val_acc": []

}


# ================================================================
# 30. FASE 1
# ================================================================

print("\n")
print("=" * 70)
print("FASE 1 - TRANSFER LEARNING")
print("=" * 70)


best_val_loss = float(
    "inf"
)


for epoch in range(

    INITIAL_EPOCHS

):


    print("\n")

    print(

        f"Época "
        f"{epoch + 1}/"
        f"{INITIAL_EPOCHS}"

    )


    train_loss, train_acc = (

        train_one_epoch(

            model,

            train_loader,

            criterion,

            optimizer,

            device

        )

    )


    val_loss, val_acc, _, _ = (

        validate(

            model,

            val_loader,

            criterion,

            device

        )

    )


    history[
        "train_loss"
    ].append(
        train_loss
    )


    history[
        "train_acc"
    ].append(
        train_acc
    )


    history[
        "val_loss"
    ].append(
        val_loss
    )


    history[
        "val_acc"
    ].append(
        val_acc
    )


    print(

        f"Train Loss: "
        f"{train_loss:.4f} | "

        f"Train Acc: "
        f"{train_acc:.4f} | "

        f"Val Loss: "
        f"{val_loss:.4f} | "

        f"Val Acc: "
        f"{val_acc:.4f}"

    )


    if val_loss < best_val_loss:

        best_val_loss = val_loss


        torch.save(

            model.state_dict(),

            MODEL_PATH

        )


        print(
            "✓ Mejor modelo guardado."
        )


# ================================================================
# 31. CARGAR MEJOR MODELO
# ================================================================

model.load_state_dict(

    torch.load(

        MODEL_PATH,

        map_location=device

    )

)


# ================================================================
# 32. FASE 2 - FINE TUNING
# ================================================================

print("\n")
print("=" * 70)
print("FASE 2 - FINE-TUNING")
print("=" * 70)


# ================================================================
# Descongelar todo
# ================================================================

for param in model.parameters():

    param.requires_grad = True


# ================================================================
# Congelar primeras capas
# ================================================================

# EfficientNet-B0:
#
# model.features contiene los bloques
#
# Entrenaremos los últimos bloques
# y el clasificador.


for param in model.features.parameters():

    param.requires_grad = False


# Descongelar los últimos bloques
#
# features[-2] y features[-1]
#

for param in model.features[-2:].parameters():

    param.requires_grad = True


# Mantener classifier entrenable

for param in model.classifier.parameters():

    param.requires_grad = True


# ================================================================
# 33. NUEVO OPTIMIZADOR
# ================================================================

optimizer = optim.Adam(

    filter(

        lambda p:
        p.requires_grad,

        model.parameters()

    ),

    lr=FINE_TUNE_LR,

    weight_decay=WEIGHT_DECAY

)


# ================================================================
# 34. FINE-TUNING
# ================================================================

best_val_loss = float(
    "inf"
)


for epoch in range(

    FINE_TUNE_EPOCHS

):


    print("\n")

    print(

        f"Fine-tuning "
        f"Época "
        f"{epoch + 1}/"
        f"{FINE_TUNE_EPOCHS}"

    )


    train_loss, train_acc = (

        train_one_epoch(

            model,

            train_loader,

            criterion,

            optimizer,

            device

        )

    )


    val_loss, val_acc, _, _ = (

        validate(

            model,

            val_loader,

            criterion,

            device

        )

    )


    history[
        "train_loss"
    ].append(
        train_loss
    )


    history[
        "train_acc"
    ].append(
        train_acc
    )


    history[
        "val_loss"
    ].append(
        val_loss
    )


    history[
        "val_acc"
    ].append(
        val_acc
    )


    print(

        f"Train Loss: "
        f"{train_loss:.4f} | "

        f"Train Acc: "
        f"{train_acc:.4f} | "

        f"Val Loss: "
        f"{val_loss:.4f} | "

        f"Val Acc: "
        f"{val_acc:.4f}"

    )


    if val_loss < best_val_loss:

        best_val_loss = val_loss


        torch.save(

            model.state_dict(),

            MODEL_PATH

        )


        print(
            "✓ Mejor modelo guardado."
        )


# ================================================================
# 35. CARGAR MEJOR MODELO
# ================================================================

model.load_state_dict(

    torch.load(

        MODEL_PATH,

        map_location=device

    )

)


model = model.to(
    device
)


# ================================================================
# 36. EVALUACIÓN FINAL
# ================================================================

print("\n")
print("=" * 70)
print("EVALUACIÓN EN TEST")
print("=" * 70)


test_loss, test_acc, _, _ = (

    validate(

        model,

        test_loader,

        criterion,

        device

    )

)


print(
    f"\nTest Loss: "
    f"{test_loss:.4f}"
)

print(
    f"Test Accuracy: "
    f"{test_acc:.4f}"
)


# ================================================================
# 37. PREDICCIONES
# ================================================================

model.eval()


all_labels = []

all_predictions = []

all_probabilities = []


with torch.no_grad():


    for images, labels in tqdm(

        test_loader,

        desc="Generando predicciones"

    ):


        images = images.to(
            device
        )


        outputs = model(
            images
        )


        probabilities = torch.softmax(

            outputs,

            dim=1

        )


        predictions = torch.argmax(

            probabilities,

            dim=1

        )


        all_labels.extend(

            labels.numpy()

        )


        all_predictions.extend(

            predictions.cpu().numpy()

        )


        all_probabilities.extend(

            probabilities.cpu().numpy()

        )


y_true = np.array(
    all_labels
)


y_pred = np.array(
    all_predictions
)


y_prob = np.array(
    all_probabilities
)


# ================================================================
# 38. MÉTRICAS
# ================================================================

accuracy = accuracy_score(

    y_true,

    y_pred

)


precision = precision_score(

    y_true,

    y_pred,

    average="weighted",

    zero_division=0

)


recall = recall_score(

    y_true,

    y_pred,

    average="weighted",

    zero_division=0

)


f1_weighted = f1_score(

    y_true,

    y_pred,

    average="weighted",

    zero_division=0

)


f1_macro = f1_score(

    y_true,

    y_pred,

    average="macro",

    zero_division=0

)


print("\n")
print("=" * 70)
print("MÉTRICAS FINALES")
print("=" * 70)


print(

    f"Accuracy          : "
    f"{accuracy:.4f}"

)


print(

    f"Precision weighted : "
    f"{precision:.4f}"

)


print(

    f"Recall weighted    : "
    f"{recall:.4f}"

)


print(

    f"F1 weighted        : "
    f"{f1_weighted:.4f}"

)


print(

    f"F1 macro           : "
    f"{f1_macro:.4f}"

)


# ================================================================
# 39. AUC ROC
# ================================================================

y_true_onehot = np.eye(

    NUM_CLASSES

)[

    y_true

]


try:

    auc_macro = roc_auc_score(

        y_true_onehot,

        y_prob,

        multi_class="ovr",

        average="macro"

    )


    auc_weighted = roc_auc_score(

        y_true_onehot,

        y_prob,

        multi_class="ovr",

        average="weighted"

    )


    print(

        f"AUC-ROC Macro      : "
        f"{auc_macro:.4f}"

    )


    print(

        f"AUC-ROC Weighted   : "
        f"{auc_weighted:.4f}"

    )


except Exception as e:

    auc_macro = np.nan

    auc_weighted = np.nan


    print(

        "No se pudo calcular AUC:",
        e

    )


# ================================================================
# 40. REPORTE DE CLASIFICACIÓN
# ================================================================

print("\n")
print("=" * 70)
print("REPORTE DE CLASIFICACIÓN")
print("=" * 70)


print(

    classification_report(

        y_true,

        y_pred,

        target_names=class_names,

        zero_division=0

    )

)


# ================================================================
# 41. MATRIZ DE CONFUSIÓN
# ================================================================

cm = confusion_matrix(

    y_true,

    y_pred

)


plt.figure(

    figsize=(9, 7)

)


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=class_names,

    yticklabels=class_names

)


plt.title(

    "Matriz de Confusión - EfficientNet-B0"

)

plt.xlabel(
    "Predicción"
)

plt.ylabel(
    "Real"
)

plt.tight_layout()

plt.show()


# ================================================================
# 42. SENSIBILIDAD Y ESPECIFICIDAD
# ================================================================

print("\n")
print("=" * 70)
print("SENSIBILIDAD Y ESPECIFICIDAD")
print("=" * 70)


sensitivities = []

specificities = []


for i in range(

    NUM_CLASSES

):


    TP = cm[i, i]


    FN = (

        np.sum(

            cm[i, :]

        )

        - TP

    )


    FP = (

        np.sum(

            cm[:, i]

        )

        - TP

    )


    TN = (

        np.sum(cm)

        - TP

        - FN

        - FP

    )


    sensitivity = (

        TP /

        (TP + FN)

        if (

            TP + FN

        ) > 0

        else 0

    )


    specificity = (

        TN /

        (TN + FP)

        if (

            TN + FP

        ) > 0

        else 0

    )


    sensitivities.append(

        sensitivity

    )


    specificities.append(

        specificity

    )


    print(

        f"\nClase {i}: "
        f"{class_names[i]}"

    )


    print(

        f"Sensibilidad: "
        f"{sensitivity:.4f}"

    )


    print(

        f"Especificidad: "
        f"{specificity:.4f}"

    )


mean_sensitivity = np.mean(

    sensitivities

)


mean_specificity = np.mean(

    specificities

)


print("\n")


print(

    f"Sensibilidad media: "
    f"{mean_sensitivity:.4f}"

)


print(

    f"Especificidad media: "
    f"{mean_specificity:.4f}"

)


# ================================================================
# 43. CURVAS DE ENTRENAMIENTO
# ================================================================

epochs = range(

    1,

    len(

        history[
            "train_loss"
        ]

    ) + 1

)


plt.figure(

    figsize=(15, 6)

)


# ================================================================
# ACCURACY
# ================================================================

plt.subplot(

    1,

    2,

    1

)


plt.plot(

    epochs,

    history[
        "train_acc"
    ],

    label="Train",

    linewidth=2

)


plt.plot(

    epochs,

    history[
        "val_acc"
    ],

    label="Validation",

    linewidth=2

)


plt.axvline(

    x=INITIAL_EPOCHS,

    color="red",

    linestyle="--",

    label="Inicio Fine-tuning"

)


plt.title(

    "Accuracy - EfficientNet-B0"

)


plt.xlabel(
    "Época"
)


plt.ylabel(
    "Accuracy"
)


plt.legend()


plt.grid(
    alpha=0.3
)


# ================================================================
# LOSS
# ================================================================

plt.subplot(

    1,

    2,

    2

)


plt.plot(

    epochs,

    history[
        "train_loss"
    ],

    label="Train",

    linewidth=2

)


plt.plot(

    epochs,

    history[
        "val_loss"
    ],

    label="Validation",

    linewidth=2

)


plt.axvline(

    x=INITIAL_EPOCHS,

    color="red",

    linestyle="--",

    label="Inicio Fine-tuning"

)


plt.title(

    "Loss - EfficientNet-B0"

)


plt.xlabel(
    "Época"
)


plt.ylabel(
    "Loss"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()

plt.show()


# ================================================================
# 44. GUARDAR RESULTADOS
# ================================================================

results = pd.DataFrame({

    "Modelo": [

        "EfficientNet-B0"

    ],

    "Accuracy": [

        accuracy

    ],

    "Precision_Weighted": [

        precision

    ],

    "Recall_Weighted": [

        recall

    ],

    "F1_Weighted": [

        f1_weighted

    ],

    "F1_Macro": [

        f1_macro

    ],

    "Sensitivity_Mean": [

        mean_sensitivity

    ],

    "Specificity_Mean": [

        mean_specificity

    ],

    "AUC_Macro": [

        auc_macro

    ],

    "AUC_Weighted": [

        auc_weighted

    ]

})


results.to_csv(

    "resultados_efficientnet_b0_aptos.csv",

    index=False

)


print("\n")
print(
    "✓ Resultados guardados en:"
)

print(
    "resultados_efficientnet_b0_aptos.csv"
)


# ================================================================
# 45. GUARDAR MODELO COMPLETO
# ================================================================

FINAL_MODEL = (

    "EfficientNet_B0_APTOS2019_M4_MPS.pth"

)


torch.save(

    {

        "model_state_dict":
        model.state_dict(),

        "class_names":
        class_names,

        "num_classes":
        NUM_CLASSES,

        "img_size":
        IMG_SIZE,

        "architecture":
        "EfficientNet-B0",

        "device":
        str(device)

    },

    FINAL_MODEL

)


print(
    "\n✓ Modelo guardado:"
)

print(
    FINAL_MODEL
)


# ================================================================
# 46. FUNCIÓN DE PREDICCIÓN
# ================================================================

def predict_image(

    image_path,

    model,

    device

):


    # ------------------------------------------------
    # Verificar archivo
    # ------------------------------------------------

    if not os.path.exists(

        image_path

    ):

        print(

            "No existe:",
            image_path

        )

        return None


    # ------------------------------------------------
    # Transformación
    # ------------------------------------------------

    transform = transforms.Compose([

        transforms.Resize(

            (IMG_SIZE, IMG_SIZE)

        ),

        transforms.ToTensor(),

        transforms.Normalize(

            mean=[

                0.485,

                0.456,

                0.406

            ],

            std=[

                0.229,

                0.224,

                0.225

            ]

        )

    ])


    # ------------------------------------------------
    # Abrir imagen
    # ------------------------------------------------

    image = Image.open(

        image_path

    ).convert(
        "RGB"
    )


    # ------------------------------------------------
    # Transformar
    # ------------------------------------------------

    image_tensor = transform(

        image

    )


    image_tensor = (

        image_tensor.unsqueeze(
            0
        )

    )


    image_tensor = image_tensor.to(

        device

    )


    # ------------------------------------------------
    # Predicción
    # ------------------------------------------------

    model.eval()


    with torch.no_grad():


        output = model(

            image_tensor

        )


        probabilities = torch.softmax(

            output,

            dim=1

        )[0]


        predicted_class = torch.argmax(

            probabilities

        ).item()


    # ------------------------------------------------
    # Mostrar resultado
    # ------------------------------------------------

    print("\n")
    print("=" * 70)

    print(
        "PREDICCIÓN"
    )

    print("=" * 70)


    print(

        f"Clase: "
        f"{predicted_class}"

    )


    print(

        f"Diagnóstico: "
        f"{class_names[predicted_class]}"

    )


    print(
        "\nProbabilidades:"
    )


    for i in range(

        NUM_CLASSES

    ):


        print(

            f"{i} - "
            f"{class_names[i]:20s}: "
            f"{probabilities[i].item() * 100:.2f}%"

        )


    # ------------------------------------------------
    # Mostrar imagen
    # ------------------------------------------------

    plt.figure(

        figsize=(7, 7)

    )


    plt.imshow(
        image
    )


    plt.axis(
        "off"
    )


    plt.title(

        f"{class_names[predicted_class]} "
        f"({probabilities[predicted_class].item() * 100:.2f}%)"

    )


    plt.show()


    return (

        predicted_class,

        probabilities.cpu().numpy()

    )


# ================================================================
# 47. EJEMPLO DE PREDICCIÓN
# ================================================================

example_image = (

    test_df.iloc[0][
        "image_path"
    ]

)


predict_image(

    example_image,

    model,

    device

)


# ================================================================
# 48. RESUMEN FINAL
# ================================================================

print("\n")
print("=" * 70)
print("RESUMEN FINAL")
print("=" * 70)


print(

    f"Device              : "
    f"{device}"

)


print(

    f"Modelo              : "
    f"EfficientNet-B0"

)


print(

    f"Dataset             : "
    f"APTOS 2019"

)


print(

    f"Clases              : "
    f"{NUM_CLASSES}"

)


print(

    f"Tamaño imagen       : "
    f"{IMG_SIZE} x {IMG_SIZE}"

)


print(

    f"Batch size          : "
    f"{BATCH_SIZE}"

)


print(

    f"Accuracy            : "
    f"{accuracy:.4f}"

)


print(

    f"Precision weighted  : "
    f"{precision:.4f}"

)


print(

    f"Recall weighted     : "
    f"{recall:.4f}"

)


print(

    f"F1 weighted         : "
    f"{f1_weighted:.4f}"

)


print(

    f"F1 macro            : "
    f"{f1_macro:.4f}"

)


print(

    f"Sensibilidad media  : "
    f"{mean_sensitivity:.4f}"

)


print(

    f"Especificidad media : "
    f"{mean_specificity:.4f}"

)


print(

    f"AUC-ROC Macro       : "
    f"{auc_macro:.4f}"

)


print(

    f"AUC-ROC Weighted    : "
    f"{auc_weighted:.4f}"

)


print("=" * 70)

print(
    "ENTRENAMIENTO FINALIZADO"
)

print("=" * 70)
```
