```python
# ================================================================
# VISION TRANSFORMER (ViT) + PYTORCH + MPS
# APTOS 2019 BLINDNESS DETECTION
# APPLE SILICON M4
# ================================================================
#
# CLASIFICACIÓN DE RETINOPATÍA DIABÉTICA
#
# CLASES:
# 0 = No DR
# 1 = Mild
# 2 = Moderate
# 3 = Severe
# 4 = Proliferative DR
#
# MODELO:
# Vision Transformer (ViT)
#
# ================================================================


# ================================================================
# 1. IMPORTACIONES
# ================================================================

import os
import random

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image

from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader

from torchvision import transforms

from sklearn.model_selection import train_test_split

from sklearn.utils.class_weight import compute_class_weight

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)


# ================================================================
# 2. CONFIGURACIÓN
# ================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)


# ================================================================
# 3. DEVICE - APPLE M4
# ================================================================

print("=" * 70)
print("CONFIGURACIÓN DEL SISTEMA")
print("=" * 70)

print(
    "Versión de PyTorch:",
    torch.__version__
)


if torch.backends.mps.is_available():

    device = torch.device("mps")

    print("✓ Apple Silicon MPS disponible.")

elif torch.cuda.is_available():

    device = torch.device("cuda")

    print("✓ CUDA disponible.")

else:

    device = torch.device("cpu")

    print("⚠ MPS/CUDA no disponible.")
    print("Se utilizará CPU.")


print(
    "Device seleccionado:",
    device
)


# ================================================================
# 4. RUTA DEL DATASET
# ================================================================

# MODIFICAR ESTA RUTA

BASE_DIR = (
   "./aptos2019-blindness-detection"
)


CSV_PATH = os.path.join(
    BASE_DIR,
    "train.csv"
)


IMAGE_DIR = os.path.join(
    BASE_DIR,
    "train_images"
)


# ================================================================
# 5. HIPERPARÁMETROS
# ================================================================

IMG_SIZE = 224

PATCH_SIZE = 16

BATCH_SIZE = 16

NUM_CLASSES = 5

NUM_WORKERS = 0

EPOCHS = 10

LEARNING_RATE = 0.00005

WEIGHT_DECAY = 0.0001


MODEL_PATH = (
    "best_vit_aptos_m4_mps.pth"
)


FINAL_MODEL = (
    "ViT_APTOS2019_M4_MPS.pth"
)


# ================================================================
# 6. CONFIGURACIÓN DEL ViT
# ================================================================

EMBED_DIM = 256

NUM_HEADS = 8

NUM_LAYERS = 6

MLP_DIM = 512

DROPOUT = 0.1


# ================================================================
# 7. NOMBRES DE CLASE
# ================================================================

class_names = [

    "No DR",

    "Mild",

    "Moderate",

    "Severe",

    "Proliferative DR"

]


# ================================================================
# 8. COMPROBAR DATASET
# ================================================================

print("\n")
print("=" * 70)
print("COMPROBANDO DATASET")
print("=" * 70)


if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(

        f"No se encontró:\n{CSV_PATH}\n\n"
        f"Modifica BASE_DIR."

    )


if not os.path.exists(IMAGE_DIR):

    raise FileNotFoundError(

        f"No se encontró:\n{IMAGE_DIR}\n\n"
        f"Modifica BASE_DIR."

    )


print(
    "✓ Dataset encontrado."
)


# ================================================================
# 9. CARGAR CSV
# ================================================================

df = pd.read_csv(
    CSV_PATH
)


print("\n")
print("=" * 70)
print("INFORMACIÓN DEL DATASET")
print("=" * 70)


print(
    "Número de imágenes:",
    len(df)
)


print(
    "Columnas:",
    df.columns.tolist()
)


print(
    "\nPrimeras filas:"
)


print(
    df.head()
)


# ================================================================
# 10. CREAR RUTAS
# ================================================================

df["image_path"] = df[
    "id_code"
].apply(

    lambda x:

    os.path.join(

        IMAGE_DIR,

        str(x) + ".png"

    )

)


# ================================================================
# 11. COMPROBAR IMÁGENES
# ================================================================

df["exists"] = df[
    "image_path"
].apply(
    os.path.exists
)


print("\n")


print(
    "Imágenes encontradas:",
    df["exists"].sum()
)


print(
    "Imágenes faltantes:",
    (~df["exists"]).sum()
)


df = df[
    df["exists"]
].copy()


df.reset_index(
    drop=True,
    inplace=True
)


# ================================================================
# 12. DISTRIBUCIÓN DE CLASES
# ================================================================

print("\n")
print("=" * 70)
print("DISTRIBUCIÓN DE CLASES")
print("=" * 70)


distribution = (

    df["diagnosis"]
    .value_counts()
    .sort_index()

)


for class_id, count in distribution.items():

    print(

        f"Clase {class_id} - "
        f"{class_names[class_id]}: "
        f"{count}"

    )


# ================================================================
# 13. GRÁFICA DE DISTRIBUCIÓN
# ================================================================

plt.figure(
    figsize=(10, 6)
)


sns.countplot(

    data=df,

    x="diagnosis",

    hue="diagnosis",

    palette="viridis",

    legend=False

)


plt.title(
    "Distribución de clases - APTOS 2019"
)


plt.xlabel(
    "Grado de retinopatía diabética"
)


plt.ylabel(
    "Número de imágenes"
)


plt.xticks(

    ticks=[0, 1, 2, 3, 4],

    labels=[

        "0 - No DR",

        "1 - Mild",

        "2 - Moderate",

        "3 - Severe",

        "4 - Proliferative"

    ]

)


plt.tight_layout()

plt.show()


# ================================================================
# 14. TRAIN / VALIDATION / TEST
# ================================================================

train_df, temp_df = train_test_split(

    df,

    test_size=0.30,

    stratify=df["diagnosis"],

    random_state=SEED

)


val_df, test_df = train_test_split(

    temp_df,

    test_size=0.50,

    stratify=temp_df["diagnosis"],

    random_state=SEED

)


print("\n")
print("=" * 70)
print("DIVISIÓN DEL DATASET")
print("=" * 70)


print(
    f"Train      : {len(train_df)}"
)


print(
    f"Validation : {len(val_df)}"
)


print(
    f"Test       : {len(test_df)}"
)


# ================================================================
# 15. DATASET PERSONALIZADO
# ================================================================

class APTOSDataset(Dataset):


    def __init__(
        self,
        dataframe,
        transform=None
    ):

        self.dataframe = (

            dataframe.reset_index(
                drop=True
            )

        )

        self.transform = transform


    def __len__(
        self
    ):

        return len(
            self.dataframe
        )


    def __getitem__(
        self,
        index
    ):

        row = self.dataframe.iloc[
            index
        ]


        image_path = row[
            "image_path"
        ]


        label = int(
            row[
                "diagnosis"
            ]
        )


        image = Image.open(

            image_path

        ).convert(
            "RGB"
        )


        if self.transform:

            image = self.transform(
                image
            )


        return image, label


# ================================================================
# 16. TRANSFORMACIONES
# ================================================================

train_transform = transforms.Compose([

    transforms.Resize(
        (IMG_SIZE, IMG_SIZE)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        degrees=10
    ),

    transforms.RandomResizedCrop(

        IMG_SIZE,

        scale=(0.90, 1.0)

    ),

    transforms.ColorJitter(

        brightness=0.10,

        contrast=0.10

    ),

    transforms.ToTensor(),

    transforms.Normalize(

        mean=[

            0.485,
            0.456,
            0.406

        ],

        std=[

            0.229,
            0.224,
            0.225

        ]

    )

])


test_transform = transforms.Compose([

    transforms.Resize(
        (IMG_SIZE, IMG_SIZE)
    ),

    transforms.ToTensor(),

    transforms.Normalize(

        mean=[

            0.485,
            0.456,
            0.406

        ],

        std=[

            0.229,
            0.224,
            0.225

        ]

    )

])


# ================================================================
# 17. DATASETS
# ================================================================

train_dataset = APTOSDataset(

    train_df,

    transform=train_transform

)


val_dataset = APTOSDataset(

    val_df,

    transform=test_transform

)


test_dataset = APTOSDataset(

    test_df,

    transform=test_transform

)


# ================================================================
# 18. DATALOADERS
# ================================================================

train_loader = DataLoader(

    train_dataset,

    batch_size=BATCH_SIZE,

    shuffle=True,

    num_workers=NUM_WORKERS,

    pin_memory=False

)


val_loader = DataLoader(

    val_dataset,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=False

)


test_loader = DataLoader(

    test_dataset,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=False

)


print(
    "\n✓ DataLoaders creados."
)


# ================================================================
# 19. CLASS WEIGHTS
# ================================================================

classes = np.array(

    sorted(

        train_df[
            "diagnosis"
        ].unique()

    )

)


class_weights_np = compute_class_weight(

    class_weight="balanced",

    classes=classes,

    y=train_df[
        "diagnosis"
    ]

)


class_weights = torch.tensor(

    class_weights_np,

    dtype=torch.float32

).to(device)


print("\n")
print("=" * 70)
print("CLASS WEIGHTS")
print("=" * 70)


for i, weight in enumerate(
    class_weights_np
):

    print(

        f"{i} - "
        f"{class_names[i]:20s}: "
        f"{weight:.4f}"

    )


# ================================================================
# 20. MLP DEL TRANSFORMER
# ================================================================

class MLPBlock(nn.Module):


    def __init__(

        self,

        embed_dim,

        mlp_dim,

        dropout=0.1

    ):

        super().__init__()


        self.network = nn.Sequential(

            nn.Linear(

                embed_dim,

                mlp_dim

            ),

            nn.GELU(),

            nn.Dropout(
                dropout
            ),

            nn.Linear(

                mlp_dim,

                embed_dim

            ),

            nn.Dropout(
                dropout
            )

        )


    def forward(self, x):

        return self.network(x)


# ================================================================
# 21. BLOQUE ENCODER TRANSFORMER
# ================================================================

class TransformerEncoderBlock(nn.Module):


    def __init__(

        self,

        embed_dim,

        num_heads,

        mlp_dim,

        dropout=0.1

    ):

        super().__init__()


        self.norm1 = nn.LayerNorm(

            embed_dim

        )


        self.attention = nn.MultiheadAttention(

            embed_dim=embed_dim,

            num_heads=num_heads,

            dropout=dropout,

            batch_first=True

        )


        self.norm2 = nn.LayerNorm(

            embed_dim

        )


        self.mlp = MLPBlock(

            embed_dim,

            mlp_dim,

            dropout

        )


    def forward(self, x):


        # --------------------------------------------------------
        # SELF ATTENTION
        # --------------------------------------------------------

        normalized = self.norm1(
            x
        )


        attention_output, _ = (

            self.attention(

                normalized,

                normalized,

                normalized

            )

        )


        x = x + attention_output


        # --------------------------------------------------------
        # MLP
        # --------------------------------------------------------

        x = x + self.mlp(

            self.norm2(x)

        )


        return x


# ================================================================
# 22. VISION TRANSFORMER
# ================================================================

class VisionTransformer(nn.Module):


    def __init__(

        self,

        image_size=224,

        patch_size=16,

        num_classes=5,

        embed_dim=256,

        num_heads=8,

        num_layers=6,

        mlp_dim=512,

        dropout=0.1

    ):

        super().__init__()


        assert (

            image_size % patch_size == 0

        ), (

            "image_size debe ser divisible "
            "por patch_size."

        )


        # --------------------------------------------------------
        # NÚMERO DE PATCHES
        # --------------------------------------------------------

        num_patches = (

            image_size // patch_size

        ) ** 2


        self.num_patches = num_patches


        # --------------------------------------------------------
        # PATCH EMBEDDING
        # --------------------------------------------------------
        #
        # Conv2D realiza:
        #
        # Imagen → patches → embeddings
        #

        self.patch_embedding = nn.Conv2d(

            in_channels=3,

            out_channels=embed_dim,

            kernel_size=patch_size,

            stride=patch_size

        )


        # --------------------------------------------------------
        # CLASS TOKEN
        # --------------------------------------------------------

        self.cls_token = nn.Parameter(

            torch.zeros(

                1,

                1,

                embed_dim

            )

        )


        # --------------------------------------------------------
        # POSITION EMBEDDING
        # --------------------------------------------------------

        self.position_embedding = nn.Parameter(

            torch.zeros(

                1,

                num_patches + 1,

                embed_dim

            )

        )


        self.dropout = nn.Dropout(

            dropout

        )


        # --------------------------------------------------------
        # TRANSFORMER ENCODERS
        # --------------------------------------------------------

        self.transformer_layers = nn.ModuleList([

            TransformerEncoderBlock(

                embed_dim=embed_dim,

                num_heads=num_heads,

                mlp_dim=mlp_dim,

                dropout=dropout

            )

            for _ in range(num_layers)

        ])


        # --------------------------------------------------------
        # NORMALIZACIÓN FINAL
        # --------------------------------------------------------

        self.norm = nn.LayerNorm(

            embed_dim

        )


        # --------------------------------------------------------
        # CLASIFICADOR
        # --------------------------------------------------------

        self.head = nn.Sequential(

            nn.Linear(

                embed_dim,

                128

            ),

            nn.GELU(),

            nn.Dropout(
                0.3
            ),

            nn.Linear(

                128,

                num_classes

            )

        )


        # --------------------------------------------------------
        # INICIALIZACIÓN
        # --------------------------------------------------------

        nn.init.trunc_normal_(

            self.cls_token,

            std=0.02

        )


        nn.init.trunc_normal_(

            self.position_embedding,

            std=0.02

        )


    def forward(self, x):


        # --------------------------------------------------------
        # PATCH EMBEDDING
        # --------------------------------------------------------

        x = self.patch_embedding(x)


        # Shape:
        #
        # [B, embed_dim, H/patch, W/patch]


        # --------------------------------------------------------
        # CONVERTIR A SECUENCIA
        # --------------------------------------------------------

        x = x.flatten(

            2

        )


        # [B, embed_dim, patches]


        x = x.transpose(

            1,

            2

        )


        # [B, patches, embed_dim]


        # --------------------------------------------------------
        # CLASS TOKEN
        # --------------------------------------------------------

        batch_size = x.shape[0]


        cls_tokens = self.cls_token.expand(

            batch_size,

            -1,

            -1

        )


        x = torch.cat(

            (

                cls_tokens,

                x

            ),

            dim=1

        )


        # --------------------------------------------------------
        # POSITION EMBEDDING
        # --------------------------------------------------------

        x = (

            x

            +

            self.position_embedding

        )


        x = self.dropout(
            x
        )


        # --------------------------------------------------------
        # TRANSFORMER
        # --------------------------------------------------------

        for layer in self.transformer_layers:

            x = layer(x)


        # --------------------------------------------------------
        # NORMALIZACIÓN
        # --------------------------------------------------------

        x = self.norm(x)


        # --------------------------------------------------------
        # CLASS TOKEN
        # --------------------------------------------------------

        cls_output = x[:, 0]


        # --------------------------------------------------------
        # CLASIFICACIÓN
        # --------------------------------------------------------

        output = self.head(

            cls_output

        )


        return output


# ================================================================
# 23. CREAR MODELO
# ================================================================

print("\n")
print("=" * 70)
print("CREANDO VISION TRANSFORMER")
print("=" * 70)


model = VisionTransformer(

    image_size=IMG_SIZE,

    patch_size=PATCH_SIZE,

    num_classes=NUM_CLASSES,

    embed_dim=EMBED_DIM,

    num_heads=NUM_HEADS,

    num_layers=NUM_LAYERS,

    mlp_dim=MLP_DIM,

    dropout=DROPOUT

)


model = model.to(
    device
)


print(
    model
)


# ================================================================
# 24. PARÁMETROS
# ================================================================

total_params = sum(

    p.numel()

    for p in model.parameters()

)


trainable_params = sum(

    p.numel()

    for p in model.parameters()

    if p.requires_grad

)


print("\n")


print(

    f"Parámetros totales: "
    f"{total_params:,}"

)


print(

    f"Parámetros entrenables: "
    f"{trainable_params:,}"

)


# ================================================================
# 25. FUNCIÓN DE PÉRDIDA
# ================================================================

criterion = nn.CrossEntropyLoss(

    weight=class_weights

)


# ================================================================
# 26. OPTIMIZADOR
# ================================================================

optimizer = optim.AdamW(

    model.parameters(),

    lr=LEARNING_RATE,

    weight_decay=WEIGHT_DECAY

)


# ================================================================
# 27. SCHEDULER
# ================================================================

scheduler = optim.lr_scheduler.ReduceLROnPlateau(

    optimizer,

    mode="min",

    factor=0.5,

    patience=3

)


# ================================================================
# 28. TRAIN ONE EPOCH
# ================================================================

def train_one_epoch(

    model,

    loader,

    criterion,

    optimizer,

    device

):


    model.train()


    running_loss = 0.0

    correct = 0

    total = 0


    progress_bar = tqdm(

        loader,

        desc="Training",

        leave=False

    )


    for images, labels in progress_bar:


        images = images.to(
            device
        )


        labels = labels.to(
            device
        )


        # --------------------------------------------------------
        # RESET GRADIENTS
        # --------------------------------------------------------

        optimizer.zero_grad()


        # --------------------------------------------------------
        # FORWARD
        # --------------------------------------------------------

        outputs = model(
            images
        )


        # --------------------------------------------------------
        # LOSS
        # --------------------------------------------------------

        loss = criterion(

            outputs,

            labels

        )


        # --------------------------------------------------------
        # BACKPROPAGATION
        # --------------------------------------------------------

        loss.backward()


        # --------------------------------------------------------
        # OPTIMIZER
        # --------------------------------------------------------

        optimizer.step()


        # --------------------------------------------------------
        # MÉTRICAS
        # --------------------------------------------------------

        running_loss += (

            loss.item()

            *

            images.size(0)

        )


        _, predicted = torch.max(

            outputs,

            1

        )


        total += labels.size(0)


        correct += (

            predicted == labels

        ).sum().item()


        progress_bar.set_postfix(

            loss=f"{loss.item():.4f}"

        )


    epoch_loss = (

        running_loss /

        total

    )


    epoch_accuracy = (

        correct /

        total

    )


    return (

        epoch_loss,

        epoch_accuracy

    )


# ================================================================
# 29. VALIDACIÓN
# ================================================================

def validate(

    model,

    loader,

    criterion,

    device

):


    model.eval()


    running_loss = 0.0

    correct = 0

    total = 0


    all_labels = []

    all_predictions = []


    with torch.no_grad():


        for images, labels in loader:


            images = images.to(
                device
            )


            labels = labels.to(
                device
            )


            outputs = model(
                images
            )


            loss = criterion(

                outputs,

                labels

            )


            running_loss += (

                loss.item()

                *

                images.size(0)

            )


            _, predicted = torch.max(

                outputs,

                1

            )


            total += labels.size(0)


            correct += (

                predicted == labels

            ).sum().item()


            all_labels.extend(

                labels.cpu().numpy()

            )


            all_predictions.extend(

                predicted.cpu().numpy()

            )


    epoch_loss = (

        running_loss /

        total

    )


    epoch_accuracy = (

        correct /

        total

    )


    return (

        epoch_loss,

        epoch_accuracy,

        np.array(all_labels),

        np.array(all_predictions)

    )


# ================================================================
# 30. HISTORIAL
# ================================================================

history = {

    "train_loss": [],

    "train_acc": [],

    "val_loss": [],

    "val_acc": [],

    "learning_rate": []

}


# ================================================================
# 31. ENTRENAMIENTO
# ================================================================

print("\n")
print("=" * 70)
print("INICIANDO ENTRENAMIENTO ViT")
print("=" * 70)


best_val_loss = float(
    "inf"
)


best_val_acc = 0.0


patience = 7


epochs_without_improvement = 0


for epoch in range(
    EPOCHS
):


    print("\n")


    print(

        f"ÉPOCA "
        f"{epoch + 1}/{EPOCHS}"

    )


    # ------------------------------------------------------------
    # TRAIN
    # ------------------------------------------------------------

    train_loss, train_acc = (

        train_one_epoch(

            model,

            train_loader,

            criterion,

            optimizer,

            device

        )

    )


    # ------------------------------------------------------------
    # VALIDATION
    # ------------------------------------------------------------

    (

        val_loss,

        val_acc,

        _,

        _

    ) = validate(

        model,

        val_loader,

        criterion,

        device

    )


    # ------------------------------------------------------------
    # SCHEDULER
    # ------------------------------------------------------------

    scheduler.step(
        val_loss
    )


    current_lr = optimizer.param_groups[0][
        "lr"
    ]


    # ------------------------------------------------------------
    # HISTORIAL
    # ------------------------------------------------------------

    history[
        "train_loss"
    ].append(
        train_loss
    )


    history[
        "train_acc"
    ].append(
        train_acc
    )


    history[
        "val_loss"
    ].append(
        val_loss
    )


    history[
        "val_acc"
    ].append(
        val_acc
    )


    history[
        "learning_rate"
    ].append(
        current_lr
    )


    # ------------------------------------------------------------
    # RESULTADOS
    # ------------------------------------------------------------

    print(

        f"Train Loss       : "
        f"{train_loss:.4f}"

    )


    print(

        f"Train Accuracy   : "
        f"{train_acc:.4f}"

    )


    print(

        f"Validation Loss  : "
        f"{val_loss:.4f}"

    )


    print(

        f"Validation Acc.  : "
        f"{val_acc:.4f}"

    )


    print(

        f"Learning Rate    : "
        f"{current_lr:.8f}"

    )


    # ------------------------------------------------------------
    # GUARDAR MEJOR MODELO
    # ------------------------------------------------------------

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        best_val_acc = val_acc

        epochs_without_improvement = 0


        torch.save(

            model.state_dict(),

            MODEL_PATH

        )


        print(
            "✓ Mejor modelo guardado."
        )


    else:

        epochs_without_improvement += 1


    # ------------------------------------------------------------
    # EARLY STOPPING
    # ------------------------------------------------------------

    if (

        epochs_without_improvement
        >= patience

    ):

        print("\n")

        print(
            "Early stopping activado."
        )

        break


# ================================================================
# 32. CARGAR MEJOR MODELO
# ================================================================

print("\n")
print("=" * 70)
print("CARGANDO MEJOR MODELO")
print("=" * 70)


model.load_state_dict(

    torch.load(

        MODEL_PATH,

        map_location=device

    )

)


model = model.to(
    device
)


print(
    "✓ Mejor modelo cargado."
)


# ================================================================
# 33. EVALUACIÓN EN TEST
# ================================================================

print("\n")
print("=" * 70)
print("EVALUACIÓN FINAL EN TEST")
print("=" * 70)


test_loss, test_acc, _, _ = (

    validate(

        model,

        test_loader,

        criterion,

        device

    )

)


print(

    f"Test Loss     : "
    f"{test_loss:.4f}"

)


print(

    f"Test Accuracy : "
    f"{test_acc:.4f}"

)


# ================================================================
# 34. PREDICCIONES
# ================================================================

print("\n")
print(
    "Generando predicciones..."
)


model.eval()


all_labels = []

all_predictions = []

all_probabilities = []


with torch.no_grad():


    for images, labels in tqdm(

        test_loader,

        desc="Test"

    ):


        images = images.to(
            device
        )


        outputs = model(
            images
        )


        probabilities = torch.softmax(

            outputs,

            dim=1

        )


        predictions = torch.argmax(

            probabilities,

            dim=1

        )


        all_labels.extend(

            labels.numpy()

        )


        all_predictions.extend(

            predictions.cpu().numpy()

        )


        all_probabilities.extend(

            probabilities.cpu().numpy()

        )


y_true = np.array(
    all_labels
)


y_pred = np.array(
    all_predictions
)


y_prob = np.array(
    all_probabilities
)


# ================================================================
# 35. MÉTRICAS
# ================================================================

accuracy = accuracy_score(

    y_true,

    y_pred

)


precision = precision_score(

    y_true,

    y_pred,

    average="weighted",

    zero_division=0

)


recall = recall_score(

    y_true,

    y_pred,

    average="weighted",

    zero_division=0

)


f1_weighted = f1_score(

    y_true,

    y_pred,

    average="weighted",

    zero_division=0

)


f1_macro = f1_score(

    y_true,

    y_pred,

    average="macro",

    zero_division=0

)


# ================================================================
# 36. AUC
# ================================================================

y_true_onehot = np.eye(

    NUM_CLASSES

)[
    y_true
]


try:

    auc_macro = roc_auc_score(

        y_true_onehot,

        y_prob,

        multi_class="ovr",

        average="macro"

    )


    auc_weighted = roc_auc_score(

        y_true_onehot,

        y_prob,

        multi_class="ovr",

        average="weighted"

    )


except Exception:

    auc_macro = np.nan

    auc_weighted = np.nan


# ================================================================
# 37. MATRIZ DE CONFUSIÓN
# ================================================================

cm = confusion_matrix(

    y_true,

    y_pred

)


# ================================================================
# 38. SENSIBILIDAD / ESPECIFICIDAD
# ================================================================

sensitivities = []

specificities = []


for i in range(

    NUM_CLASSES

):


    TP = cm[i, i]


    FN = (

        np.sum(
            cm[i, :]
        )

        - TP

    )


    FP = (

        np.sum(
            cm[:, i]
        )

        - TP

    )


    TN = (

        np.sum(cm)

        - TP

        - FN

        - FP

    )


    sensitivity = (

        TP /

        (TP + FN)

        if (

            TP + FN

        ) > 0

        else 0

    )


    specificity = (

        TN /

        (TN + FP)

        if (

            TN + FP

        ) > 0

        else 0

    )


    sensitivities.append(

        sensitivity

    )


    specificities.append(

        specificity

    )


mean_sensitivity = np.mean(

    sensitivities

)


mean_specificity = np.mean(

    specificities

)


# ================================================================
# 39. MÉTRICAS FINALES
# ================================================================

print("\n")
print("=" * 70)
print("MÉTRICAS FINALES - ViT")
print("=" * 70)


print(

    f"Accuracy            : "
    f"{accuracy:.4f}"

)


print(

    f"Precision Weighted  : "
    f"{precision:.4f}"

)


print(

    f"Recall Weighted     : "
    f"{recall:.4f}"

)


print(

    f"F1 Weighted         : "
    f"{f1_weighted:.4f}"

)


print(

    f"F1 Macro            : "
    f"{f1_macro:.4f}"

)


print(

    f"AUC Macro           : "
    f"{auc_macro:.4f}"

)


print(

    f"AUC Weighted        : "
    f"{auc_weighted:.4f}"

)


print(

    f"Sensibilidad Media  : "
    f"{mean_sensitivity:.4f}"

)


print(

    f"Especificidad Media : "
    f"{mean_specificity:.4f}"

)


# ================================================================
# 40. CLASSIFICATION REPORT
# ================================================================

print("\n")
print("=" * 70)
print("REPORTE DE CLASIFICACIÓN")
print("=" * 70)


print(

    classification_report(

        y_true,

        y_pred,

        target_names=class_names,

        zero_division=0

    )

)


# ================================================================
# 41. MATRIZ DE CONFUSIÓN
# ================================================================

plt.figure(

    figsize=(9, 7)

)


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=class_names,

    yticklabels=class_names

)


plt.title(

    "Matriz de Confusión - Vision Transformer"

)


plt.xlabel(
    "Predicción"
)


plt.ylabel(
    "Clase real"
)


plt.tight_layout()

plt.show()


# ================================================================
# 42. CURVAS DE ENTRENAMIENTO
# ================================================================

epochs_completed = range(

    1,

    len(

        history[
            "train_loss"
        ]

    ) + 1

)


plt.figure(

    figsize=(15, 6)

)


# ------------------------------------------------
# ACCURACY
# ------------------------------------------------

plt.subplot(

    1,

    2,

    1

)


plt.plot(

    epochs_completed,

    history[
        "train_acc"
    ],

    label="Train",

    linewidth=2

)


plt.plot(

    epochs_completed,

    history[
        "val_acc"
    ],

    label="Validation",

    linewidth=2

)


plt.title(

    "Accuracy - Vision Transformer"

)


plt.xlabel(
    "Época"
)


plt.ylabel(
    "Accuracy"
)


plt.legend()


plt.grid(
    alpha=0.3
)


# ------------------------------------------------
# LOSS
# ------------------------------------------------

plt.subplot(

    1,

    2,

    2

)


plt.plot(

    epochs_completed,

    history[
        "train_loss"
    ],

    label="Train",

    linewidth=2

)


plt.plot(

    epochs_completed,

    history[
        "val_loss"
    ],

    label="Validation",

    linewidth=2

)


plt.title(

    "Loss - Vision Transformer"

)


plt.xlabel(
    "Época"
)


plt.ylabel(
    "Loss"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()

plt.show()


# ================================================================
# 43. LEARNING RATE
# ================================================================

plt.figure(

    figsize=(8, 5)

)


plt.plot(

    epochs_completed,

    history[
        "learning_rate"
    ],

    marker="o"

)


plt.title(

    "Learning Rate - ViT"

)


plt.xlabel(
    "Época"
)


plt.ylabel(
    "Learning Rate"
)


plt.yscale(
    "log"
)


plt.grid(
    alpha=0.3
)


plt.tight_layout()

plt.show()


# ================================================================
# 44. RESULTADOS POR CLASE
# ================================================================

class_results = pd.DataFrame({

    "Clase": class_names,

    "Sensitivity": sensitivities,

    "Specificity": specificities

})


print("\n")
print("=" * 70)
print("RESULTADOS POR CLASE")
print("=" * 70)


print(
    class_results
)


# ================================================================
# 45. GUARDAR RESULTADOS
# ================================================================

results = pd.DataFrame({

    "Modelo": [

        "Vision Transformer (ViT)"

    ],

    "Accuracy": [

        accuracy

    ],

    "Precision_Weighted": [

        precision

    ],

    "Recall_Weighted": [

        recall

    ],

    "F1_Weighted": [

        f1_weighted

    ],

    "F1_Macro": [

        f1_macro

    ],

    "Sensitivity_Mean": [

        mean_sensitivity

    ],

    "Specificity_Mean": [

        mean_specificity

    ],

    "AUC_Macro": [

        auc_macro

    ],

    "AUC_Weighted": [

        auc_weighted

    ]

})


results.to_csv(

    "resultados_vit_aptos.csv",

    index=False

)


print("\n")


print(
    "✓ Resultados guardados:"
)


print(
    "resultados_vit_aptos.csv"
)


# ================================================================
# 46. GUARDAR MODELO
# ================================================================

torch.save(

    {

        "model_state_dict":
        model.state_dict(),

        "class_names":
        class_names,

        "num_classes":
        NUM_CLASSES,

        "img_size":
        IMG_SIZE,

        "patch_size":
        PATCH_SIZE,

        "embed_dim":
        EMBED_DIM,

        "num_heads":
        NUM_HEADS,

        "num_layers":
        NUM_LAYERS,

        "mlp_dim":
        MLP_DIM,

        "architecture":
        "Vision Transformer",

        "device":
        str(device)

    },

    FINAL_MODEL

)


print("\n")


print(
    "✓ Modelo guardado:"
)


print(
    FINAL_MODEL
)


# ================================================================
# 47. FUNCIÓN DE PREDICCIÓN
# ================================================================

def predict_image(

    image_path,

    model,

    device

):


    if not os.path.exists(

        image_path

    ):

        print(

            "No existe la imagen:",

            image_path

        )

        return None


    transform = transforms.Compose([

        transforms.Resize(

            (IMG_SIZE, IMG_SIZE)

        ),

        transforms.ToTensor(),

        transforms.Normalize(

            mean=[

                0.485,

                0.456,

                0.406

            ],

            std=[

                0.229,

                0.224,

                0.225

            ]

        )

    ])


    image = Image.open(

        image_path

    ).convert(
        "RGB"
    )


    image_tensor = transform(

        image

    )


    image_tensor = (

        image_tensor

        .unsqueeze(0)

        .to(device)

    )


    model.eval()


    with torch.no_grad():


        output = model(

            image_tensor

        )


        probabilities = torch.softmax(

            output,

            dim=1

        )[0]


        predicted_class = torch.argmax(

            probabilities

        ).item()


    print("\n")
    print("=" * 70)
    print("PREDICCIÓN")
    print("=" * 70)


    print(

        f"Clase: "
        f"{predicted_class}"

    )


    print(

        f"Diagnóstico: "
        f"{class_names[predicted_class]}"

    )


    print("\nProbabilidades:")


    for i in range(

        NUM_CLASSES

    ):


        print(

            f"{i} - "
            f"{class_names[i]:20s}: "
            f"{probabilities[i].item() * 100:.2f}%"

        )


    # ------------------------------------------------------------
    # MOSTRAR IMAGEN
    # ------------------------------------------------------------

    plt.figure(

        figsize=(7, 7)

    )


    plt.imshow(
        image
    )


    plt.axis(
        "off"
    )


    plt.title(

        f"{class_names[predicted_class]} "
        f"("
        f"{probabilities[predicted_class].item() * 100:.2f}%"
        f")"

    )


    plt.show()


    return (

        predicted_class,

        probabilities.cpu().numpy()

    )


# ================================================================
# 48. EJEMPLO DE PREDICCIÓN
# ================================================================

example_image = (

    test_df.iloc[0][
        "image_path"
    ]

)


predict_image(

    example_image,

    model,

    device

)


# ================================================================
# 49. RESUMEN FINAL
# ================================================================

print("\n")
print("=" * 70)
print("RESUMEN FINAL")
print("=" * 70)


print(

    f"Device              : "
    f"{device}"

)


print(

    f"Modelo              : "
    f"Vision Transformer"

)


print(

    f"Dataset             : "
    f"APTOS 2019"

)


print(

    f"Clases              : "
    f"{NUM_CLASSES}"

)


print(

    f"Tamaño imagen       : "
    f"{IMG_SIZE} x {IMG_SIZE}"

)


print(

    f"Patch size           : "
    f"{PATCH_SIZE} x {PATCH_SIZE}"

)


print(

    f"Embedding dimension : "
    f"{EMBED_DIM}"

)


print(

    f"Attention heads      : "
    f"{NUM_HEADS}"

)


print(

    f"Transformer layers   : "
    f"{NUM_LAYERS}"

)


print(

    f"Batch size           : "
    f"{BATCH_SIZE}"

)


print(

    f"Épocas ejecutadas    : "
    f"{len(epochs_completed)}"

)


print(

    f"Accuracy             : "
    f"{accuracy:.4f}"

)


print(

    f"Precision Weighted   : "
    f"{precision:.4f}"

)


print(

    f"Recall Weighted      : "
    f"{recall:.4f}"

)


print(

    f"F1 Weighted          : "
    f"{f1_weighted:.4f}"

)


print(

    f"F1 Macro             : "
    f"{f1_macro:.4f}"

)


print(

    f"AUC Macro            : "
    f"{auc_macro:.4f}"

)


print(

    f"Sensibilidad Media   : "
    f"{mean_sensitivity:.4f}"

)


print(

    f"Especificidad Media  : "
    f"{mean_specificity:.4f}"

)


print("=" * 70)


print(
    "ENTRENAMIENTO FINALIZADO"
)


print("=" * 70)
```
